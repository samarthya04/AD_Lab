1. Write a Python program to demonstrate data preprocessing steps: handling missing values, encoding categorical data, and feature scaling.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
from sklearn.impute import SimpleImputer
```

```python
data = {
    'Age': [22, np.nan, 24, 30, 25],
    'Salary': [50000, 60000, np.nan, 70000, 80000],
    'Country': ['USA', 'Canada', 'India', 'Canada', np.nan],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female']
}
```

```python
df = pd.DataFrame(data)
print(df)
print(df.isnull())
print(df.isnull().sum())
```

```python
imputer_num = SimpleImputer(strategy='mean')
df[['Age', 'Salary']] = imputer_num.fit_transform(df[['Age', 'Salary']])
imputer_cat = SimpleImputer(strategy='most_frequent')
df[['Country']] = imputer_cat.fit_transform(df[['Country']])
print(df)
```

```python
df.describe()
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
label_encoder = LabelEncoder()
df['Gender'] = label_encoder.fit_transform(df['Gender'])
df['Country'] = label_encoder.fit_transform(df['Country'])
print(df)
```

```python
scaler = MinMaxScaler()
df[['Age', 'Salary']] = scaler.fit_transform(df[['Age', 'Salary']])
print(df)
```

2. Load a sample dataset using pandas (e.g., Iris and Wine datasets).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
cols = ["sepal length in cm", "sepal width in cm", "petal length in cm", "petal width in cm", "species"]
data = pd.read_csv("iris.data", names=cols)
data.head()
```

```python
data.describe()
```

```python
data.info()
```

3.    Plot the distribution of a feature using matplotlib.pyplot.hist().

```python
plt.figure(figsize=(8, 6))
plt.hist(data[['sepal length in cm']], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length in cm')
plt.ylabel('Frequency')
plt.show()
```

4.    Create scatter plots to understand relationships between features using seaborn.scatterplot().

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=data, x=data['sepal length in cm'], y=data['sepal width in cm'], hue='species', palette='viridis')
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length in cm')
plt.ylabel('Sepal Width in cm')
plt.show()
```

5.    Use a correlation heatmap to find the relationship between multiple features with seaborn.heatmap().


```python
correlation_matrix = data.iloc[:, :-1].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Features')
plt.show()
```
