```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('/content/house_prices.csv')

square_feet = df.iloc[:, :-1].values.flatten()
prices = df.iloc[:, -1].values
x_mean = np.mean(square_feet)
y_mean = np.mean(prices)

n = len(square_feet)
numerator = np.sum((square_feet - x_mean) * (prices - y_mean))
denominator = np.sum((square_feet - x_mean) ** 2)

m = numerator / denominator
b = y_mean - (m * x_mean)

print(f"Linear Regression Equation: Price = {m:.2f} * Square Footage + {b:.2f}")

predictions = m * square_feet + b
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv('/content/house_prices.csv')

square_feet = df.iloc[:, :-1].values
prices = df.iloc[:, -1].values

model = LinearRegression()
model.fit(square_feet, prices)

m = model.coef_[0]
b = model.intercept_

print(f"Linear Regression Equation: Price = {m:.2f} * Square Footage + {b:.2f}")

predictions = model.predict(square_feet)
```

```python
import numpy as np

X = np.array([
    [4, 11],
    [8, 4],
    [13, 5],
    [7, 14]
])

mean = np.mean(X, axis=0)
X_centered = X - mean

cov_matrix = np.cov(X_centered.T, bias=False)

eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

idx = eigenvalues.argsort()[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

principal_component = eigenvectors[:, 0]

projected_data = np.dot(X_centered, principal_component)

print("Reduced dataset:\n", projected_data)
```

```python
import numpy as np

X = np.array([[2, 10], [2, 5], [8, 4], [5, 8], [7, 5], [6, 4], [1, 2], [4, 9]])
centers = np.array([[2, 10], [5, 8], [1, 2]])

for _ in range(2):
    distances = np.linalg.norm(X[:, np.newaxis] - centers, axis=2)
    labels = np.argmin(distances, axis=1)
    centers = np.array([X[labels == k].mean(axis=0) for k in range(3)])
    print("Centers in each iteration:\n", centers)

print("Final clusters:", labels)
```
